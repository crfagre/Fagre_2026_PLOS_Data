```python
import pysam
import matplotlib.pyplot as plt
from collections import Counter

def plot_end_distribution(bam_path, ref_name, end_type="5prime", start_pos=None, end_pos=None):
    # Open the BAM file (must be indexed)
    bam = pysam.AlignmentFile(bam_path, "rb")
    
    positions = []
    
    # Fetch reads mapping to your specific reference
    for read in bam.fetch(ref_name, start_pos, end_pos):
        if read.is_unmapped:
            continue
            
        if end_type == "5prime":
            # Forward strand 5' is the start; Reverse strand 5' is the end
            pos = read.reference_start if not read.is_reverse else read.reference_end
        elif end_type == "3prime":
            # Forward strand 3' is the end; Reverse strand 3' is the start
            pos = read.reference_end if not read.is_reverse else read.reference_start
        else:
            raise ValueError("end_type must be '5prime' or '3prime'")
            
        positions.append(pos)
        
    bam.close()
    
    # Count frequencies of each position
    pos_counts = Counter(positions)
    
    # Sort and fill in zero-counts for continuous plotting/shading
    if pos_counts:
        min_pos = min(pos_counts.keys())
        max_pos = max(pos_counts.keys())
        full_positions = list(range(min_pos, max_pos + 1))
        full_counts = [pos_counts.get(p, 0) for p in full_positions]
    else:
        full_positions, full_counts = [], []
    
    # Plotting
    plt.figure(figsize=(12, 5))
    
    if full_positions:
        # 1. Draw the line
        plt.plot(full_positions, full_counts, color='royalblue', linestyle='-')
        
        # 2. Add the light shading underneath
        # alpha=0.2 makes it 20% opaque (lightly shaded)
        plt.fill_between(full_positions, full_counts, color='royalblue', alpha=0.2)
    
    plt.title(f"{end_type} Read End Distribution for {ref_name}")
    plt.xlabel("Genomic Coordinate")
    plt.ylabel("Read Count")
    
    # Optional: limit x-axis to your region of interest
    if start_pos and end_pos:
        plt.xlim(start_pos, end_pos)
        
    plt.tight_layout()
    plt.show()


```

```python
import pysam
import matplotlib.pyplot as plt
from collections import Counter

def plot_fragment_abundance_barplot(bam_path, ref_name, start, end, top_n=None):
    # 1. Open BAM and extract all reads in the window
    bam = pysam.AlignmentFile(bam_path, "rb")
    all_reads = []
    
    for read in bam.fetch(ref_name, start, end):
        if read.is_unmapped:
            continue
        # Extract 5p (start) and 3p (end) coordinates
        five_p = read.reference_start
        three_p = read.reference_end
        all_reads.append((five_p, three_p))
    bam.close()
    
    if not all_reads:
        print("No reads found in this region.")
        return

    # 2. Calculate abundance and sort by frequency (high to low)
    # Counter.most_common() automatically sorts from highest to lowest count
    read_counts = Counter(all_reads)
    sorted_fragments = read_counts.most_common()
    
    # Optional: Limit to top N fragments to avoid overcrowding the x-axis
    if top_n is not None:
        sorted_fragments = sorted_fragments[:top_n]

    # 3. Prepare data for plotting
    # Format the labels exactly as [5p, 3p]
    labels = [f"[{frag[0]}, {frag[1]}]" for frag, count in sorted_fragments]
    counts = [count for frag, count in sorted_fragments]

    # 4. Plotting
    plt.figure(figsize=(12, 6))
    
    # Create the barplot
    plt.bar(labels, counts, color='royalblue', edgecolor='black')
    
    plt.xlabel("Fragment Coordinates [5p, 3p]")
    plt.ylabel("Read Count (Abundance)")
    
    title_suffix = f" (Top {top_n})" if top_n else ""
    plt.title(f"Fragment Abundance in {ref_name}:{start}-{end}{title_suffix}")
    
    # Rotate x-axis labels by 45 degrees so they don't overlap
    plt.xticks(rotation=45, ha='right')
    
    # Add a grid on the y-axis to make it easier to read values
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    
    plt.tight_layout()
    plt.show()

# --- Example Usage ---
# Plot all fragments:
# plot_fragment_abundance_barplot("your_data.bam", "ref_1", start=0, end=100)

# Plot only the top 30 most abundant fragments:
# plot_fragment_abundance_barplot("your_data.bam", "ref_1", start=0, end=100, top_n=30)
```

```python
import pysam
import matplotlib.pyplot as plt
import random

def plot_split_sorted_reads(bam_path, ref_name, start, end, sample_size=200, output_filename=None):
    # 1. Open BAM and extract all reads in the window
    bam = pysam.AlignmentFile(bam_path, "rb")
    all_reads = []
    
    for read in bam.fetch(ref_name, start, end):
        if read.is_unmapped:
            continue
        all_reads.append((read.reference_start, read.reference_end))
    bam.close()
    
    if not all_reads:
        print("No reads found in this region.")
        return

    # 2. Randomly subset the reads
    if len(all_reads) > sample_size:
        sampled_reads = random.sample(all_reads, sample_size)
    else:
        sampled_reads = all_reads

    # 3. Calculate window midpoint to determine "Left" vs "Right" side
    window_midpoint = start + (end - start) / 2
    
    left_reads = []
    right_reads = []
    
    # 4. Assign reads to the left or right half
    for s, e in sampled_reads:
        read_midpoint = s + (e - s) / 2
        if read_midpoint < window_midpoint:
            left_reads.append((s, e))
        else:
            right_reads.append((s, e))

    # 5. Sort the two groups independently
    left_reads.sort(key=lambda x: (x[0], x[1]))
    right_reads.sort(key=lambda x: (-x[1], x[0]))

    # 6. Plotting
    plt.close('all') 
    
    # Using a high DPI (300) ensures the exported JPG is print/publication quality
    plt.figure(figsize=(10, 5), dpi=300) 
    
    # Plot Left Reads using barh
    if left_reads:
        l_starts = [x[0] for x in left_reads]
        l_widths = [x[1] - x[0] for x in left_reads]
        l_y = range(len(left_reads))
        plt.barh(y=l_y, width=l_widths, left=l_starts, height=1.0, color='slategray', edgecolor='none', antialiased=False)

    # Plot Right Reads using barh
    if right_reads:
        r_starts = [x[0] for x in right_reads]
        r_widths = [x[1] - x[0] for x in right_reads]
        r_y = range(len(right_reads))
        plt.barh(y=r_y, width=r_widths, left=r_starts, height=1.0, color='slategray', edgecolor='none', antialiased=False)
    
    # Add a dashed line down the exact center
    plt.axvline(x=window_midpoint, color='black', linestyle='--', alpha=0.3)
    
    plt.xlabel("Coordinate Position")
    plt.ylabel("Individual Reads")
    plt.title(f"Alignment of {len(sampled_reads)} Reads\n(Left and Right groups both start at y=0)")
    
    plt.yticks([]) 
    plt.xlim(start, end)
    plt.tight_layout()
    
    # 7. Export the plot to JPG
    if output_filename:
        # Enforce the .jpg extension if it was forgotten
        if not output_filename.lower().endswith(('.jpg', '.jpeg')):
            output_filename += '.jpg'
            
        # bbox_inches='tight' prevents the axis labels from being cut off in the exported file
        plt.savefig(output_filename, format='jpg', dpi=300, bbox_inches='tight')
        print(f"Exported plot to {output_filename}")
        
    plt.show()

# --- Example Usage ---
# plot_split_sorted_reads("your_data.bam", "ref_1", start=0, end=100, sample_size=200, output_filename="read_alignment_plot")
```

```python
plot_split_sorted_reads("shEmpty_1_NT.mtrna.sort.bam", "tRNA-Val-AAC-4", start=0, end=76, sample_size = 200)
```
