##Fig 1c read coverage of mtrnas

```python
import pysam
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.collections import LineCollection

def plot_reads_with_deletions(bam_path, chrom, output_file="figures/WT_1_NT_del.png"):
    bam = pysam.AlignmentFile(bam_path, "rb")
    
    grey_segments = []
    red_segments = []  # To store the deletion coordinates
    rows = [] 
    all_read_positions = []
    read_records = []
    
    for read in bam.fetch(chrom):
        # We need a CIGAR string to find deletions
        if read.is_unmapped or not read.cigartuples: 
            continue
            
        r_start = read.reference_start
        r_end = read.reference_end
        all_read_positions.append((r_start, r_end))
        read_records.append((chrom, r_start, r_end))
        
        # 1. Packing Algorithm (Find the row)
        placed = False
        assigned_row = -1
        for i, row_end in enumerate(rows):
            if r_start > row_end: 
                rows[i] = r_end
                assigned_row = i
                placed = True
                break
        
        if not placed:
            rows.append(r_end)
            assigned_row = len(rows) - 1
            
        # Add the main read backbone
        grey_segments.append(((r_start, assigned_row), (r_end, assigned_row)))
        
        # 2. Parse CIGAR for Deletions
        # CIGAR codes: 0=Match, 1=Insertion, 2=Deletion, 3=Skip, 4=SoftClip
        current_ref_pos = r_start
        for op, length in read.cigartuples:
            if op in (0, 7, 8):  # Matches (consume reference bases)
                current_ref_pos += length
            elif op == 2:        # Deletion (D)
                del_start = current_ref_pos
                del_end = current_ref_pos + length
                # Record the deletion segment on the same row
                red_segments.append(((del_start, assigned_row), (del_end, assigned_row)))
                current_ref_pos += length
            elif op == 3:        # Reference skip (N)
                current_ref_pos += length
            # Insertions (1) and Soft clips (4) do not advance the reference position

    bam.close()

    if not grey_segments:
        print(f"No reads found for {chrom}.")
        return

    # Write read coordinates to CSV
    csv_path = output_file.rsplit(".", 1)[0] + "_reads.csv"
    pd.DataFrame(read_records, columns=["ref", "start", "end"]).to_csv(csv_path, index=False)

    # Calculate bounds for X-axis limit
    start_pos = min(p[0] for p in all_read_positions)
    end_pos = max(p[1] for p in all_read_positions)

    # 3. Visual Configuration
    font_size = 20
    fig, ax = plt.subplots(figsize=(14, 4)) 

    # Plot Read Segments (Grey background, zorder=1 to keep them at the bottom)
    lc_grey = LineCollection(grey_segments, colors='grey', linewidths=0.5, zorder=1)
    ax.add_collection(lc_grey)
    
    # Plot Deletion Segments (Red, zorder=2 so they draw on top of the grey lines)
    if red_segments:
        # Made linewidth slightly thicker (1.5) so single-nt deletions pop out
        lc_red = LineCollection(red_segments, colors='red', linewidths=1.5, zorder=2)
        ax.add_collection(lc_red)
    
    # Custom Labels and Formatting
    ax.set_ylabel("Reads", fontsize=font_size, labelpad=15)
    ax.set_xlabel("Position (nt)", fontsize=font_size, labelpad=15)
    
    ax.set_ylim(-1, len(rows))
    ax.set_xlim(start_pos - 1, end_pos + 1)
    
    ax.set_yticks([]) 
    ax.tick_params(axis='x', labelsize=font_size - 4)

    for spine in ['top', 'right', 'left']:
        ax.spines[spine].set_visible(False)
    
    plt.tight_layout()
    plt.savefig(output_file, dpi=300, bbox_inches='tight')
    plt.show()

# Execution:
plot_reads_with_deletions("WT_1_NT.mtrna.sort.bam", "tRNA-Lys-TTT-5")
```
